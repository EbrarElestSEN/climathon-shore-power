```python
import pandas as pd
import os

print("Çalışma klasörü:", os.getcwd())
print("\nKlasördeki dosyalar:")
print(os.listdir())

power_file = "AE_power_use_at_berth.csv"
emissions_file = "AE_emissions_at_berth.csv"

power_df = pd.read_csv(power_file)
emissions_df = pd.read_csv(emissions_file)

print("\nElektrik gücü verisi boyutu:", power_df.shape)
print("Emisyon verisi boyutu:", emissions_df.shape)

print("\nElektrik gücü verisinin ilk 5 satırı:")
display(power_df.iloc[:5, :10])

print("\nSütun adları:")
print(power_df.columns[:12].tolist())

print("\nGemi türleri:")
print(power_df["Ship category"].dropna().unique())
```

```python

import matplotlib.pyplot as plt

# İlk limanın verilerini seç
row = power_df.iloc[0]

# 2018 yılındaki ilk 168 saati (7 gün) al
hour_columns = [str(i) for i in range(1, 169)]
weekly_power = row[hour_columns].astype(float)

# Grafiği çiz
plt.figure(figsize=(12, 5))
plt.plot(range(1, 169), weekly_power.values)

plt.title(
    f"{row['Port name']} - {row['Ship category']}\n"
    "İlk 7 Günlük Saatlik Elektrik Gücü İhtiyacı"
)
plt.xlabel("Saat")
plt.ylabel("Elektrik gücü ihtiyacı (kW)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Liman:", row["Port name"])
print("Gemi kategorisi:", row["Ship category"])
print("En yüksek saatlik güç ihtiyacı:", weekly_power.max(), "kW")
print("İlk 7 gündeki ortalama güç ihtiyacı:", weekly_power.mean(), "kW")
```

```python
#4. hücre — Elektrik ihtiyacı olan bir kayıt bulma

# Saatlik sütunları belirle
hour_columns = [str(i) for i in range(1, 8761)]

# Her satır için yıl boyunca toplam elektrik gücü değerini hesapla
total_power = power_df[hour_columns].sum(axis=1)

# Toplamı sıfırdan büyük olan ilk kaydı bul
nonzero_rows = power_df.index[total_power > 0]

if len(nonzero_rows) > 0:
    row_index = nonzero_rows[0]
    row = power_df.loc[row_index]

    print("Bulunan satır:", row_index)
    print("Liman:", row["Port name"])
    print("Ülke:", row["Country abbreviation"])
    print("Gemi kategorisi:", row["Ship category"])
    print("Yıllık saatlik değerlerin toplamı:", total_power.loc[row_index])
else:
    print("Veri setinde pozitif elektrik gücü değeri bulunamadı.")
```

```python
#5. hücre — İlk pozitif elektrik ihtiyacını bulma
# Seçtiğimiz kaydın saatlik değerleri
hourly_values = row[hour_columns].astype(float)

# Elektrik ihtiyacının sıfırdan büyük olduğu ilk saat
positive_hours = hourly_values[hourly_values > 0]

print("Elektrik ihtiyacı olan saat sayısı:", len(positive_hours))
print("İlk pozitif saat:", positive_hours.index[0])
print("O saatteki elektrik gücü:", positive_hours.iloc[0], "kW")
```

```python
#6. hücre — Elektrik ihtiyacının olduğu haftayı çizme


import matplotlib.pyplot as plt

# 4207. saatin çevresindeki 168 saatlik aralığı seç
start_hour = 4207
end_hour = min(start_hour + 167, 8760)

selected_hours = [str(i) for i in range(start_hour, end_hour + 1)]
selected_power = row[selected_hours].astype(float)

plt.figure(figsize=(12, 5))
plt.plot(range(start_hour, end_hour + 1), selected_power.values)

plt.title(f"{row['Port name']} - {row['Ship category']}\nElektrik Gücü İhtiyacı")
plt.xlabel("Yıl içindeki saat")
plt.ylabel("Elektrik gücü ihtiyacı (kW)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("İncelenen saat aralığı:", start_hour, "-", end_hour)
print("En yüksek güç ihtiyacı:", selected_power.max(), "kW")
print("Pozitif değer bulunan saat sayısı:", (selected_power > 0).sum())
```

```python
#7. hücre — Genel veri analizi


# Her satır için yıllık toplam elektrik gücü değerleri
annual_totals = power_df[hour_columns].sum(axis=1)

print("Toplam kayıt sayısı:", len(power_df))
print("Liman sayısı:", power_df["Port name"].nunique())
print("Ülke sayısı:", power_df["Country abbreviation"].nunique())
print("Gemi kategorisi sayısı:", power_df["Ship category"].nunique())

print("\nYıl boyunca pozitif değeri olan kayıt sayısı:",
      (annual_totals > 0).sum())

print("Yıl boyunca tamamen sıfır olan kayıt sayısı:",
      (annual_totals == 0).sum())

print("\nEn yüksek yıllık toplam değere sahip ilk 10 kayıt:")
top10 = power_df[
    ["Port name", "Country abbreviation", "Ship category"]
].copy()

top10["Annual total"] = annual_totals
display(top10.sort_values("Annual total", ascending=False).head(10))
```

```python
#8. hücre — Saatlik verilerdeki sıfır ve pozitif değerleri inceleme


# Saatlik elektrik gücü değerlerini sayısal olarak al
hourly_data = power_df[hour_columns].astype(float)

total_values = hourly_data.size
zero_values = (hourly_data == 0).sum().sum()
positive_values = (hourly_data > 0).sum().sum()

print("Toplam saatlik veri noktası:", total_values)
print("Sıfır olan değer sayısı:", zero_values)
print("Pozitif değer sayısı:", positive_values)

print(f"\nSıfır değer oranı: %{zero_values / total_values * 100:.2f}")
print(f"Pozitif değer oranı: %{positive_values / total_values * 100:.2f}")

print("\nSaatlik güç değerlerinin özeti (kW):")
display(hourly_data.stack().describe())
```

```python
#9. hücre — Elektrik ihtiyacının dağılımını görselleştirelim


import matplotlib.pyplot as plt

# Sıfır ve pozitif değerlerin dağılımı
labels = ["Sıfır değerler", "Pozitif değerler"]
counts = [zero_values, positive_values]

plt.figure(figsize=(7, 5))
plt.bar(labels, counts)
plt.title("Saatlik Elektrik Gücü Değerlerinin Dağılımı")
plt.ylabel("Veri noktası sayısı")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# Sadece pozitif değerlerin dağılımı
positive_data = hourly_data.to_numpy().ravel()
positive_data = positive_data[positive_data > 0]

plt.figure(figsize=(8, 5))
plt.hist(positive_data, bins=50)
plt.title("Pozitif Saatlik Elektrik Gücü Değerlerinin Dağılımı")
plt.xlabel("Elektrik gücü ihtiyacı (kW)")
plt.ylabel("Frekans")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#10. hücre — İlk tahmin modelini hazırlayalım


# Rotterdam limanındaki konteyner gemisi kaydını bul
mask = (
    (power_df["Port name"] == "ROTTERDAM") &
    (power_df["Ship category"] == "Container ship")
)

rotterdam_rows = power_df[mask]

print("Bulunan kayıt sayısı:", len(rotterdam_rows))
display(rotterdam_rows[
    ["Port name", "Country abbreviation", "Ship category"]
])
```

```python
#11. hücre — Veriyi eğitim ve test olarak ayırma


from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import matplotlib.pyplot as plt

# Rotterdam konteyner gemisi kaydının 8760 saatlik verisi
rotterdam_row = rotterdam_rows.iloc[0]
series = rotterdam_row[hour_columns].astype(float).to_numpy()

# İlk %80 eğitim, son %20 test
split_index = int(len(series) * 0.8)
train = series[:split_index]
test = series[split_index:]

print("Toplam saat:", len(series))
print("Eğitim saatleri:", len(train))
print("Test saatleri:", len(test))
print("Eğitim dönemi son saat:", split_index)
print("Test dönemi saatleri:", len(test))
```

```python
#12. hücre — Basit referans tahmini

# Eğitim verisinin ortalamasını hesapla
baseline_value = train.mean()

# Her test saati için aynı tahmini kullan
baseline_predictions = np.full(len(test), baseline_value)

# Hata ölçümleri
mae = mean_absolute_error(test, baseline_predictions)
rmse = np.sqrt(mean_squared_error(test, baseline_predictions))

print("Eğitim verisi ortalaması:", round(baseline_value, 2), "kW")
print("Baseline MAE:", round(mae, 2), "kW")
print("Baseline RMSE:", round(rmse, 2), "kW")
```

```python
#13. hücre — Random Forest tahmin modeli


from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

# Geçmiş saatleri özellik olarak oluştur
data = pd.DataFrame({"power": series})
data["lag_1"] = data["power"].shift(1)
data["lag_24"] = data["power"].shift(24)
data["lag_168"] = data["power"].shift(168)

# İlk 168 saatte yeterli geçmiş veri olmadığı için bu satırları çıkar
data = data.dropna()

features = ["lag_1", "lag_24", "lag_168"]

# Eğitim ve test hedeflerini zaman sırasını bozmadan ayır
train_data = data.loc[data.index < split_index]
test_data = data.loc[data.index >= split_index]

X_train = train_data[features]
y_train = train_data["power"]

X_test = test_data[features]
y_test = test_data["power"]

# Modeli eğit
model = RandomForestRegressor(
    n_estimators=100,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

# Test dönemindeki saatlik tahminler
predictions = model.predict(X_test)

# Sonuçları hesapla
rf_mae = mean_absolute_error(y_test, predictions)
rf_rmse = np.sqrt(mean_squared_error(y_test, predictions))

print("Random Forest MAE:", round(rf_mae, 2), "kW")
print("Random Forest RMSE:", round(rf_rmse, 2), "kW")
print("Eğitim örneği sayısı:", len(X_train))
print("Test örneği sayısı:", len(X_test))
```

```python
#14. hücre — Gerçek ve tahmin edilen değerleri çizelim


import matplotlib.pyplot as plt

# Test döneminin ilk 168 saatini göster
n = min(168, len(y_test))

plt.figure(figsize=(13, 5))
plt.plot(
    y_test.iloc[:n].to_numpy(),
    label="Gerçek değer",
    linewidth=1.5
)
plt.plot(
    predictions[:n],
    label="Random Forest tahmini",
    linewidth=1.5
)

plt.title("Rotterdam Konteyner Gemileri: Gerçek ve Tahmin Edilen Güç")
plt.xlabel("Test dönemindeki saat")
plt.ylabel("Elektrik gücü ihtiyacı (kW)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#15. hücre — Gerçek ve tahmin değerlerini ayrı ayrı değerlendirelim


# Gerçek talebin pozitif olduğu test saatlerini bul
positive_mask = y_test.to_numpy() > 0

print("Test saatlerinin toplamı:", len(y_test))
print("Gerçek elektrik ihtiyacı olan saat:", positive_mask.sum())
print("Gerçek elektrik ihtiyacı sıfır olan saat:", (~positive_mask).sum())

if positive_mask.sum() > 0:
    positive_mae = mean_absolute_error(
        y_test.to_numpy()[positive_mask],
        predictions[positive_mask]
    )
    print(
        "Pozitif talep saatlerindeki MAE:",
        round(positive_mae, 2), "kW"
    )

print(
    "Gerçek değerlerin ortalaması:",
    round(y_test.mean(), 2), "kW"
)
print(
    "Tahminlerin ortalaması:",
    round(predictions.mean(), 2), "kW"
)
```

```python
#16. hücre — Tahmin hatasını yüzde olarak hesaplayalım

# Ortalama mutlak hatanın gerçek ortalamaya oranı
relative_mae = (
    mean_absolute_error(y_test, predictions) / y_test.mean()
) * 100

# Ortalama tahmin sapması
mean_error = np.mean(predictions - y_test.to_numpy())

print(f"Ortalama mutlak hata / gerçek ortalama: %{relative_mae:.2f}")
print(f"Ortalama tahmin sapması: {mean_error:.2f} kW")
```

```python
#17. hücre — Modeli başka bir liman/gemi kaydında test etme

# Yıllık toplam gücü en yüksek kayıtları sırala
annual_totals = power_df[hour_columns].sum(axis=1)
top_indices = annual_totals.sort_values(ascending=False).index[:10]

# İkinci kaydı seç
test_row_index = top_indices[1]
test_row = power_df.loc[test_row_index]
test_series = test_row[hour_columns].astype(float).to_numpy()

print("Liman:", test_row["Port name"])
print("Ülke:", test_row["Country abbreviation"])
print("Gemi kategorisi:", test_row["Ship category"])

# Gecikmeli özellikleri oluştur
other_data = pd.DataFrame({"power": test_series})
other_data["lag_1"] = other_data["power"].shift(1)
other_data["lag_24"] = other_data["power"].shift(24)
other_data["lag_168"] = other_data["power"].shift(168)
other_data = other_data.dropna()

other_train = other_data.loc[other_data.index < split_index]
other_test = other_data.loc[other_data.index >= split_index]

other_features = ["lag_1", "lag_24", "lag_168"]

other_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

other_model.fit(
    other_train[other_features],
    other_train["power"]
)

other_predictions = other_model.predict(other_test[other_features])
other_mae = mean_absolute_error(other_test["power"], other_predictions)
other_rmse = np.sqrt(
    mean_squared_error(other_test["power"], other_predictions)
)

print("MAE:", round(other_mae, 2), "kW")
print("RMSE:", round(other_rmse, 2), "kW")
```

```python
#Sıradaki adım: 18. hücre — Gerçek ve tahmin değerlerini karşılaştırma


n = min(168, len(other_test))

plt.figure(figsize=(13, 5))
plt.plot(
    other_test["power"].iloc[:n].to_numpy(),
    label="Gerçek değer",
    linewidth=1.5
)
plt.plot(
    other_predictions[:n],
    label="Random Forest tahmini",
    linewidth=1.5
)

plt.title("Antwerpen Konteyner Gemileri: Gerçek ve Tahmin Edilen Güç")
plt.xlabel("Test dönemindeki saat")
plt.ylabel("Elektrik gücü ihtiyacı (kW)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#19. hücre — İki limanın performansını karşılaştırma


comparison = pd.DataFrame({
    "Liman": ["ROTTERDAM", "ANTWERPEN"],
    "Gemi kategorisi": ["Container ship", "Container ship"],
    "MAE (kW)": [rf_mae, other_mae],
    "RMSE (kW)": [rf_rmse, other_rmse]
})

display(comparison)

# Hata değerlerini karşılaştıran grafik
comparison.set_index("Liman")[["MAE (kW)", "RMSE (kW)"]].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Random Forest Modeli: Limanlara Göre Tahmin Hataları")
plt.ylabel("Hata (kW)")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
